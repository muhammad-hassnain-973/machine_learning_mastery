# Loading Data
`Different Methods of loading and writing data using pandas`
| **Function** | **Description** |
| --- | ---|
| `read_csv` | Load delimited data from a file, URL, or file-like object; use comma as default delimiter | 
| `read_fwf ` | Read data in fixed-width column format (i.e., no delimiters)
 | 
| `read_clipboard` | Variation of read_csv that reads data from the clipboard; useful for converting tables from web



## Working with txt and csv

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("example.csv")
df
```

```python
#Loading without headers
df=pd.read_csv('example.csv', header=None)
df
```

```python
#loading with custom header
df=pd.read_csv("example.csv", names=["a", "b", "c", "d", "message"])
df
```

```python
#Suppose you wanted the message column to be the index of the returned DataFrame.
# You can either indicate you want the column at index 4 or named "message" using
# the index_col argument:
names = ["a", "b", "c", "d", "message"]
df=pd.read_csv("example.csv", names=names, index_col="message")
df
```

```python
# If you want to form a hierarchical index
parsed = pd.read_csv("example2.csv",index_col=["key1", "key2"])
parsed
```

```python
result = pd.read_csv("example.txt", sep="\s+")
result
```

```python
#Loading Data with Missing Values
result = pd.read_csv("example3.csv")
pd.isna(result)
```

```python
result = pd.read_csv("example3.csv", na_values=['NULL'])
result
```

```python
result2 = pd.read_csv("example3.csv", keep_default_na=False)
result2
```

```python
# Reading Data in small pieces
pd.read_csv("example.csv", nrows=2)
```

## Working with json data

```python
obj = """
{"name": "Wes",
 "cities_lived": ["Akron", "Nashville", "New York", "San Francisco"],
 "pet": null,
 "siblings": [{"name": "Scott", "age": 34, "hobbies": ["guitars", "soccer"]},
 {"name": "Katie", "age": 42, "hobbies": ["diving", "art"]}]
}
"""
```

```python
# Python to json
import json
result=json.loads(obj)
result
```

```python
# Json to python
asjson = json.dumps(result)
asjson
```

```python
#Reading json with pandas
df=pd.read_json("example.json")
df
```
